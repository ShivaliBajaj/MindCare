```python
# MindCare - EDA
# Goal:
# understand dataset structure
# understand behavioral patterns
# inspect suicidality-related text


# imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path


# make plots slightly bigger
plt.rcParams["figure.figsize"] = (8, 5)


# dataset path
DATA_PATH = Path("../data/raw/suicidality_dataset.csv")


# load dataset
df = pd.read_csv(# dataset path

DATA_PATH = Path("../data/raw/suicidality_dataset.csv"))


# first 5 rows
df.head()
```

```python
# dataset shape
df.shape
```

```python
# column names
df.columns
```

```python
# dataset info
df.info()
```

```python
# missing values
df.isnull().sum()
```

```python
# class distribution
df["label"].value_counts()
```

```python
# visualize class imbalance
df["label"].value_counts().plot(kind="bar")

plt.title("Class Distribution")
plt.xlabel("Label")
plt.ylabel("Count")

plt.show()
```

### Observation

Dataset shows class imbalance.

In suicide-risk detection, recall is more important than accuracy because missing a high-risk individual can be clinically dangerous.

```python
# non-suicidal example
df[df["label"] == 0]["text"].iloc[0]
```

```python
# suicidal example
df[df["label"] == 1]["text"].iloc[0]
```

### Behavioral Observation

High-risk texts contain stronger hopelessness, emotional exhaustion, and socially isolated language patterns.

Some texts also repeatedly use emotionally absolute phrasing.

```python
# create text length feature
df["text_length"] = df["text"].apply(len)
```

```python
# text length stats
df["text_length"].describe()
```

```python
# text length distribution
plt.hist(df["text_length"], bins=50)

plt.title("Text Length Distribution")
plt.xlabel("Text Length")
plt.ylabel("Frequency")

plt.show()
```

```python
# average text length per class
df.groupby("label")["text_length"].mean()
```
